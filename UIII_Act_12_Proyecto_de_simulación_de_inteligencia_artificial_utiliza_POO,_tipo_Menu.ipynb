### Diccionario conductor

```python
# --- SISTEMA DE IA: GESTIÓN INTEGRAL DE CONDUCTORES (POO TOTAL) ---

class Conductor:
    """Clase que encapsula toda la gestión y la IA de conductores."""

    def __init__(self):
        """Constructor que inicializa las estructuras solicitadas."""
        # 'conductor' es la lista global de registros
        self.conductor = []
        # 'los_conductores' es el diccionario temporal para capturar datos
        self.los_conductores = {}

    def agregar_conductor(self):
        """Captura datos y los guarda en la lista interna."""
        print("\n--- Registro de Nuevo Conductor ---")

        # Entrada de datos (Key: Value) para el diccionario solicitado
        self.los_conductores = {
            "licencia": input("Tipo de licencia (A/B/C): "),
            "ruta": input("Ruta asignada: "),
            "placas": input("Placas del vehículo: "),
            "horas": int(input("Horas de manejo acumuladas: ")),
            "empresa": input("Empresa de transporte: "),
            "record": input("Récord de seguridad (limpio/riesgo): ").lower()
        }

        # Se añade una copia del diccionario a la lista 'conductor'
        self.conductor.append(self.los_conductores.copy())
        print("IA: Registro exitoso dentro de la base de datos de la clase.")

    def listar_conductor(self):
        """Muestra la información en una tabla bien alineada y distribuida."""
        if not self.conductor:
            print("\nIA: La lista 'conductor' está vacía.")
        else:
            # Encabezado con anchos fijos para alineación profesional [3]
            header = f"{'ID':<4} | {'Empresa':<15} | {'Placas':<10} | {'Horas':<8} | {'Récord':<10}"
            print("\n" + "=" * 60)
            print(header)
            print("-" * 60)

            # Recorrido de la lista interna con ciclo for [4, 5]
            indice = 0
            for c in self.conductor:
                fila = f"{indice:<4} | {c['empresa']:<15} | {c['placas']:<10} | {c['horas']:<8} | {c['record']:<10}"
                print(fila)
                indice += 1
            print("=" * 60)

    def actualizar_conductor(self):
        """Modifica un registro existente buscando por su posición (ID)."""
        self.listar_conductor()
        if self.conductor:
            pos = int(input("\nIA: Ingrese el ID del conductor a actualizar: "))
            # Verificación de rango para evitar errores
            if 0 <= pos < len(self.conductor):
                print("Ingrese los nuevos datos de seguridad:")
                self.conductor[pos]["horas"] = int(input("Nuevas horas de manejo: "))
                self.conductor[pos]["record"] = input("Nuevo récord (limpio/riesgo): ").lower()
                print("IA: Información actualizada correctamente.")
            else:
                print("IA: ID no válido.")

    def borrar_conductor(self):
        """Elimina un registro de la lista 'conductor'."""
        self.listar_conductor()
        if self.conductor:
            pos = int(input("\nIA: Ingrese el ID del conductor a borrar: "))
            if 0 <= pos < len(self.conductor):
                eliminado = self.conductor.pop(pos) # Borrado físico [6]
                print(f"IA: Registro con placas {eliminado['placas']} eliminado.")
            else:
                print("IA: ID no válido.")

    def analizar_ia(self):
        """Simulación de IA que procesa riesgos basados en los datos."""
        if not self.conductor:
            print("\nIA: No hay datos para analizar.")
        else:
            print("\n--- INFORME DE SEGURIDAD IA ---")
            for c in self.conductor:
                # Lógica condicional para evaluar seguridad [7, 8]
                if c['horas'] > 12:
                    print(f">> Unidad [{c['placas']}]: ALERTA DE FATIGA. Detener operación.")
                elif c['record'] == "riesgo":
                    print(f">> Unidad [{c['placas']}]: REVISIÓN PRIORITARIA. Historial negativo.")
                else:
                    print(f">> Unidad [{c['placas']}]: OPERACIÓN SEGURA.")

    def ejecutar_menu(self):
        """Controlador interactivo del ciclo principal."""
        while True:
            print("\n******************************************")
            print("   MENÚ DE GESTIÓN DE CONDUCTORES (POO)")
            print("******************************************")
            print("1. Agregar Conductor")
            print("2. Listar Conductores (Tabla)")
            print("3. Actualizar Seguridad")
            print("4. Borrar Conductor")
            print("5. Ejecutar Análisis IA")
            print("6. Salir")

            opcion = input("\nEditor, seleccione una opción (1-6): ")

            if opcion == "1": self.agregar_conductor()
            elif opcion == "2": self.listar_conductor()
            elif opcion == "3": self.actualizar_conductor()
            elif opcion == "4": self.borrar_conductor()
            elif opcion == "5": self.analizar_ia()
            elif opcion == "6":
                print("IA: Apagando sistemas. ¡Buen viaje!")
                break # Rompe el ciclo while True [9]
            else:
                print("IA: Error. Opción no válida.")

# --- INICIO DEL PROGRAMA ---
# Se crea el objeto 'asistente' de la clase Conductor y se inicia su menú
asistente = Conductor()
asistente.ejecutar_menu()
```
